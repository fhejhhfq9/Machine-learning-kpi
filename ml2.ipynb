```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

nyc = pd.read_csv("C:\\Users\\user\\Desktop\\uni\\4sem\\gpvy\\data.csv")
nyc.columns = ['Date', 'Temperature', 'Anomaly']
nyc.Date =nyc.Date.floordiv(100)
nyc.head()
```

```python
sns.regplot(x="Date",y="Temperature",data=nyc,scatter=False)
plt.xlabel("Рік")
plt.ylabel("Температура")
plt.show()
```

```python
from scipy import stats


linear_reg = stats.linregress(x = nyc.Date, y = nyc.Temperature)
m = linear_reg.slope
b = linear_reg.intercept

years = [2021, 2022, 2023, 2024]

for year in years:
     temp = m * year + b
     print(f"{year} : {temp:.2f}")
```

```python
years = [1885, 1886, 1887, 1888, 1889, 1890, 1891, 1892, 1893, 1894]

for year in years:
     temp = m * year + b
     print(f"{year} : {temp:.2f}")
```

```python
plt.close()
sns.regplot(x="Date",y="Temperature",data=nyc)
plt.xlabel("Рік")
plt.ylabel("Температура")
plt.show()
```
