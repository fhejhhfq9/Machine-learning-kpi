```python
import pandas as pd

titanic = pd.read_csv("C:\\Users\\user\\Desktop\\uni\\4sem\\gpvy\\TitanicSurvival.csv")
```

```python
pd.set_option('display.precision', 2)
titanic.head()
```

```python
titanic.tail()
```

```python
titanic.columns=['name','survived','sex','age','class']
titanic.head()
```

```python
youngest = titanic.loc[titanic["age"].idxmin()]
oldest   = titanic.loc[titanic["age"].idxmax()]

print("Youngest:", youngest["name"], "—", youngest["age"], "years old")
print("Oldest:  ", oldest["name"],   "—", oldest["age"],  "years old")
```

```python
avg_age = titanic["age"].mean()
print(f"Average age: {avg_age:.2f}")
```

```python
survived_stats = titanic[titanic["survived"] == "yes"]["age"].dropna().describe()
print("\nSurvivor stats: ")
print(survived_stats)
```

```python
women_1st = titanic[(titanic["sex"] == "female") & (titanic["class"] == "1st")]
women_1st_sorted = women_1st.sort_values(by="age")
print(women_1st_sorted[["name", "age", "survived"]].head())

                    
```

```python
youngest = women_1st_sorted.loc[women_1st_sorted["age"].idxmin()]
oldest   = women_1st_sorted.loc[women_1st_sorted["age"].idxmax()]

print("Youngest:", youngest["name"], "—", youngest["age"], "years old")
print("Oldest:  ", oldest["name"],   "—", oldest["age"],  "years old")
```

```python
survived_women = women_1st[women_1st["survived"] == "yes"].shape[0]

print("Number of women who survived: ", survived_women)
```
